```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sec_a = np.array([75,65,73,68,72,67])
sec_b = np.array([90,47,43,96,93,51])
no  = np.array([1,2,3,4,5,6])
```

```python
mean = np.mean(sec_b)
```

```python
mad_a = np.sum(np.abs(sec_a - mean)) / len(sec_a)
mad_b = np.sum(np.abs(sec_b - mean)) / len(sec_b)
```

```python
mad_a,mad_b
```

```python
plt.figure(figsize=(10,3))
plt.scatter(sec_a, no,label = "sec A")
plt.scatter(sec_b, no,color = "red", label = "sec_b")
plt.plot([mean for i in range(1,7)],no,color="blue",label="Mean")
plt.legend()
plt.show()
```

# Standard Deviation

```python
np.std(sec_a), np.std(sec_b)
```

```python
np.var(sec_a), np.var(sec_b)
```
