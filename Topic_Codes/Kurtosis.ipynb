```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
practice = pd.DataFrame({
    "Mesokurtic": [67,45,50,54,42,50,50,32,60,56,44,48,55,47,48,35,56,51,53,35],
    "Leptokurtic": [55,50,49,56,50,46,49,43,53,49,48,53,45,52,44,48,20,22,78,80],
    "Platykurtic": [30,33,34,36,39,41,42,44,46,49,50,53,56,57,60,62,63,66,67,70]
})

for col in practice.columns:
    print(col, "-> Kurtosis:", round(practice[col].kurt(), 2))
```

```python
meso_k = practice["Mesokurtic"].kurt()
lepto_k = practice["Leptokurtic"].kurt()
platy_k = practice["Platykurtic"].kurt()

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.histplot(practice["Mesokurtic"], kde=True, ax=axes[0], color="#639922")
axes[0].set_title(f"Mesokurtic\nKurtosis = {meso_k:.2f}")

sns.histplot(practice["Leptokurtic"], kde=True, ax=axes[1], color="#D85A30")
axes[1].set_title(f"Leptokurtic\nKurtosis = {lepto_k:.2f}")

sns.histplot(practice["Platykurtic"], kde=True, ax=axes[2], color="#378ADD")
axes[2].set_title(f"Platykurtic\nKurtosis = {platy_k:.2f}")

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))

sns.kdeplot(practice["Mesokurtic"], label=f"Mesokurtic ({meso_k:.2f})", color="#639922")
sns.kdeplot(practice["Leptokurtic"], label=f"Leptokurtic ({lepto_k:.2f})", color="#D85A30")
sns.kdeplot(practice["Platykurtic"], label=f"Platykurtic ({platy_k:.2f})", color="#378ADD")

plt.title("Comparing Kurtosis Shapes")
plt.legend()
plt.show()
```
