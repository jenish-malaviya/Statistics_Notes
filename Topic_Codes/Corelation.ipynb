```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("E:\Statistics\CSV Files\correlation_practice_dataset.csv")
dataset
```

```python
# Only Numeric Value
dataset.isnull().sum()
```

```python
dataset.info()
```

```python
data_corr = dataset.select_dtypes(["int64"]).corr()
data_corr
```

```python
data_cov = dataset.select_dtypes(["int64"]).cov()
data_cov
```

```python
plt.figure(figsize = (4,3))
sns.heatmap(data_corr,annot=True)
plt.show()
```

```python
sns.heatmap(data_cov,annot = True)
plt.show()
```
